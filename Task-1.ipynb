## 2. Discriminative Models vs Generative Models

| Aspect        | Discriminative Models                                        | Generative Models                                                                |
| ------------- | ------------------------------------------------------------ | -------------------------------------------------------------------------------- |
| Definition    | Learn the relationship between input data and output labels. | Learn the underlying patterns or distribution of data and can generate new data. |
| Main Use      | Mainly used for classification and prediction.               | Mainly used for creating new content or generating similar data.                 |
| Example Task  | Predict whether an email is spam or not spam.                | Generate a new email or text based on a given prompt.                            |
| Example Model | Logistic Regression, Decision Tree, Random Forest            | GPT, GAN, VAE, Diffusion Model                                                   |
| Output        | Usually predicts an existing class or value.                 | Produces new data such as text, images, audio, or code.                          |

## 3. Product Description Using ChatGPT

### Prompt Used

Write an attractive and creative product description for a new Flipkart gadget called "SmartFocus AI Glasses". The glasses have a built-in AI assistant, Bluetooth connectivity, voice control, noise cancellation, a small camera for taking photos, and a battery that lasts for 12 hours. The target customers are students and young professionals. Keep the description simple, modern, and suitable for an e-commerce website. Mention the main features and benefits and make the product sound useful without making unrealistic claims.

Capture quick moments with the built-in camera, listen to your favorite music, make voice commands, and interact with your AI assistant hands-free. With up to 12 hours of battery life, the SmartFocus AI Glasses are designed for students, professionals, travelers, and anyone who enjoys smart technology.

**Key Features:**

* Built-in AI assistant
* Voice control

* Noise cancellation
* Built-in camera
* Up to 12 hours of battery life

SmartFocus AI Glasses combine technology and convenience in a single wearable, making them a useful gadget for everyday activities.

## 4. New Generative AI Feature

### Application: Text Generation

I would add an **AI Travel Planner** feature to Zomato. This feature would allow users to enter a natural language request such as, "I am visiting Ahmedabad for two days. Suggest places to eat near popular tourist spots and create a food plan under ₹2,000."

Generative AI would understand the user's request and generate a personalized food itinerary. It could suggest restaurants, meals, timings, and estimated spending based on the user's preferences and budget. The feature could also modify the plan when the user asks for changes, such as "Make it vegetarian" or "Suggest cheaper options."

This would make the app more interactive because users could communicate with it naturally instead of searching for each restaurant individually.
